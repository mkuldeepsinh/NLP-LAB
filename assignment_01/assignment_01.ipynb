```python
from datasets import load_dataset
from huggingface_hub import login



dataset = load_dataset("ai4bharat/IndicCorpV2", "indiccorp_v2", streaming=True, split="guj_Gujr")
print(dataset)
```

import re

def regex_sentence_tokenizer(text):
    # Split into sentences using a regex that looks for sentence-ending punctuation
    # followed by whitespace or end of string. This is a simplified approach.
    sentence_endings = r'(?<!\w\.\w.)(?<![A-Z][a-z]\.)(?<=\.|\?|\!)\s+'
    sentences = re.split(sentence_endings, text)
    # Filter out any empty strings that might result from splitting
    sentences = [s.strip() for s in sentences if s.strip()]
    return sentences

```python
sample_size = 100000  # Define sample_size with a default value

sample_examples = []
for i, example in enumerate(dataset):
    if i >= sample_size:
        break
    sample_examples.append(example['text'])


processed_examples = []
for text in sample_examples:
    sentences = regex_sentence_tokenizer(text)
    tokenized_sentences = []
    for sentence in sentences:
        tokens = regex_word_tokenizer(sentence)
        tokenized_sentences.append(tokens)
    processed_examples.append({'original_text': text, 'tokenized_text': tokenized_sentences})

# Display the processed examples
for example in processed_examples:
    print("Original Text:")
    print(example['original_text'])
    print("\nTokenized Text (Sentences and Words):")
    for i, sentence_tokens in enumerate(example['tokenized_text']):
        print(f"  Sentence {i+1}: {sentence_tokens}")
    print("-" * 30)
```

```python
# Initialize counters
total_sentences = 0
total_words = 0
total_characters = 0
all_tokens = []

# Iterate through the processed examples
for example in processed_examples:
    for sentence_tokens in example['tokenized_text']:
        total_sentences += 1
        total_words += len(sentence_tokens)
        for token in sentence_tokens:
            total_characters += len(token)
            all_tokens.append(token)

# Calculate statistics
average_sentence_length = total_words / total_sentences if total_sentences > 0 else 0
average_word_length = total_characters / total_words if total_words > 0 else 0
unique_tokens = set(all_tokens)
type_token_ratio = len(unique_tokens) / total_words if total_words > 0 else 0

# Display the statistics
print(f"Total number of sentences: {total_sentences}")
print(f"Total number of words: {total_words}")
print(f"Total number of characters: {total_characters}")
print(f"Average Sentence Length (words per sentence): {average_sentence_length:.2f}")
print(f"Average word length (characters per word): {average_word_length:.2f}")
print(f"Type/Token Ratio (TTR): {type_token_ratio:.2f}")
```

```python
import re

def regex_word_tokenizer(text):
    # Split into words using a regex that finds sequences of word characters
    # This is a basic word tokenizer and might need adjustment based on specific requirements
    words = re.findall(r'\b\w+\b', text)
    return words
```
